```python
import scipy.io
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# Load the MATLAB file
mat_data = scipy.io.loadmat('/Users/dduke/Desktop/RPsolution.mat')
PG0s=np.array(mat_data['PG0s'][0]);

# Plot the data
fig=plt.figure();
plt.grid();
for k in mat_data.keys():
    if 'pdf' in k:
        PG0 = PG0s[int(k.split('_')[-1])-1] # get matching PG0 from list
        print(k, PG0, mat_data[k].shape)
        plt.loglog(mat_data[k][0,:]*1e6,mat_data[k][1,:],label='PG0 = %g' % PG0);
plt.legend();
plt.xlabel(r'Bubble size ($\mu$ m)')
plt.ylim(1e2,2e7)
```
